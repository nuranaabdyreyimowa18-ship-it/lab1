## 1. Постановка задачи

**Описание данных:** 
Данный датасет представляет собой исторические данные о продажах супермаркета, собранные в трех различных филиалах в течение 3 месяцев. Таблица содержит информацию о покупках, включая идентификатор счета, филиал, город, тип клиента, пол, категорию товара, цену за единицу, количество, дату и время транзакции, а также рейтинг удовлетворенности клиентов.

**Условный заказчик:** 
Условным заказчиком данного анализа является руководство сети супермаркетов (отдел аналитики и коммерческий директор), заинтересованное в оптимизации бизнес-процессов, повышении продаж и улучшении клиентского опыта.

**Задачи интеллектуального анализа данных:**
1. **Описательная аналитика и профилирование групп:** Анализ покупательского поведения в зависимости от пола и типа клиента (например, сравнение затрат постоянных членов клуба и обычных покупателей).
2. **Прогнозирование и поиск закономерностей во времени:** Изучение динамики продаж по дням недели и часам для оптимизации работы персонала и графика поставок.
3. **Поиск аномалий:** Выявление транзакций с аномально низким рейтингом или необычно большим объемом закупок.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Загрузка данных из папки data
df = pd.read_csv('../data/SuperMarket Analysis.csv')

# Определение размера датасета
print(f"Количество строк: {df.shape[0]}")
print(f"Количество столбцов: {df.shape[1]}")

# Вывод первых 5 строк таблицы для демонстрации преподавателю
df.head()
```

```python
# Пропуски данных
print("--- Пропуски в данных ---")
print(df.isnull().sum())

# Количество полных дубликатов
print("\n--- Количество полных дубликатов ---")
print(df.duplicated().sum())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Построение графиков с исправленным названием столбца (Sales)
plt.figure(figsize=(10, 5))

# 1. График распределения рейтингов
plt.subplot(1, 2, 1)
sns.histplot(df['Rating'], kde=True, color='skyblue')
plt.title('Распределение рейтингов')

# 2. График сумм чеков по полу покупателей
plt.subplot(1, 2, 2)
sns.boxplot(data=df, x='Gender', y='Sales', palette='Set2')
plt.title('Сумма чека по полу')

plt.tight_layout()
plt.show()
```
